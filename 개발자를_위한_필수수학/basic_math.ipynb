#### 개발자를 위한 필수 수학

```python
# 예제 1.1 파이썬으로 계산하기
my_value = 2 * (3+2)**2 / 5 - 4
print(my_value)

# 예제 1.2 파이썬에서 괄호를 사용해 명확하게 표현하기
my_value = 2 * ((3+2)**2 / 5) - 4
print(my_value)
```

```python
# 선형 함수 선언하기
def f(x):
    return 2 * x + 1

x_values = [0, 1, 2, 3]

for i in x_values:
    print(f(i))
```

```python
# 심파이로 선형함수 그래프 그리기
from sympy import *

x = symbols('x')
f = 2 * x + 1
plot(f)
```

```python
# 심파이로 2차 함수 그래프 그리기
x = symbols('x')
f = x **2 + 1
plot(f)
```

```python
# 독립 변수가 두 개인 함수의 그래프 

from sympy import *
from sympy.plotting import plot3d

x, y = symbols('x y')
f = 2*x + 3*y

plot3d(f)
```

```python
# 1 ~ 5 까지의 숫자에 2를 곱한 뒤에 합계 계산하기

summation = sum([i * 2 for i in range(1, 6)])
print(summation)
```

```python
x = [1, 4, 6 ,2]
n = len(x)

summation = sum([10*x[i] for i in range(n)])
print(summation)
```

```python
# 파이썬에서 로그 함수 사용하기
from math import log

# 2를 거듭제곱해 8이 되는 지수는?
x = log(8, 2)
print(x)
```
