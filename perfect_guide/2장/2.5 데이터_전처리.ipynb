**데이터 인코딩**

- 레이블 인코딩(Label encoding)

```python
from sklearn.preprocessing import LabelEncoder

items = ['TV', '냉장고', '세탁기', '에어컨', 'TV', '냉장고','믹서','믹서']

encoder = LabelEncoder()
encoder.fit(items)

labels = encoder.transform(items)
print('인코딩 변환값:', labels)
```

```python
print('인코딩 클래스 :', encoder.classes_)
```

```python
print('디코딩 원본 값 :', encoder.inverse_transform(labels))
```

- 원-핫 인코딩(One-Hot encoding)

```python
from sklearn.preprocessing import OneHotEncoder
import numpy as np

items = ['TV', '냉장고', '세탁기', '에어컨', 'TV', '냉장고','믹서','믹서']

items = np.array(items).reshape(-1,1) 

oh_encode = OneHotEncoder()
oh_encode.fit(items)
oh_labels = oh_encode.transform(items)

print('원-핫 인코딩 변환값:\n', oh_labels.toarray())
print('원-핫 인코딩 데이터 차원:\n', oh_labels.shape)
```

```python
import pandas as pd

df = pd.DataFrame({'item': items.flatten()})

df
```

```python
ohe_df = pd.get_dummies(df).astype('int')
ohe_df
```

**피처 스케일링과 정규화**

- StandardScaler

```python
from sklearn.datasets import load_iris
import pandas as pd

iris = load_iris()

iris_data = iris.data
iris_df = pd.DataFrame(data=iris_data, columns=iris.feature_names)

print(iris_df.mean()) # 평균

print(iris_df.var()) # 분산
```

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()
scaler.fit(iris_df)
iris_scaled = scaler.transform(iris_df)
iris_scaled_df = pd.DataFrame(data=iris_scaled, columns=iris.feature_names)

print(iris_scaled_df.mean()) # 평균
print(iris_scaled_df.var()) # 분산
```

```python
scaler = StandardScaler()
scaler.fit_transform(iris_df)
```

- MinMaxScaler

```python
from sklearn.preprocessing import MinMaxScaler

scaler = MinMaxScaler()
scaler.fit(iris_df)
iris_scaled = scaler.transform(iris_df)
iris_scaled_df = pd.DataFrame(data=iris_scaled, columns=iris.feature_names)

print(iris_scaled_df.min()) # 최소값
print(iris_scaled_df.max()) # 최대값
```

```python
scaler = MinMaxScaler()
scaler_df = scaler.fit_transform(iris_df)

print(scaler_df.min()) # 최소값
print(scaler_df.max()) # 최대값
```

- Scaler 를 이용하여 학습 데이터와 테스트 데이터에 fit(), transform(), fit_transform() 적용 시 유의 사항.

```python
from sklearn.preprocessing import MinMaxScaler
import numpy as np

train_array = np.arange(11).reshape(-1,1)
test_array = np.arange(0,6).reshape(-1,1)
```

```python
train_array, test_array
```

```python
scaler = MinMaxScaler()

scaler.fit(train_array)

train_scaled = scaler.transform(train_array)

print('원본 train_array 데이터 : ', np.round(train_array.reshape(-1), 2 ))
print('스케일링된 train_array 데이터 : ', np.round(train_scaled.reshape(-1), 2 ))

test_scaled = scaler.transform(test_array)

print('원본 test_array 데이터 : ', np.round(test_array.reshape(-1), 2 ))
print('스케일링된 test_array 데이터 : ', np.round(test_scaled.reshape(-1), 2 ))
```

```python
scaler.fit(test_array)
test_scaled = scaler.transform(test_array)

print('원본 test_array 데이터 : ', np.round(test_array.reshape(-1), 2 ))
print('스케일링된 test_array 데이터 : ', np.round(test_scaled.reshape(-1), 2 ))
```
