```python
import pandas as pd
import numpy as np
import warnings; warnings.filterwarnings("ignore")

movies = pd.read_csv(f"{path}/tmdb_5000_movies.csv")
print(movies.shape)
movies.head()
```

```python
movie_df = movies[['id', 'title', 'genres', 'vote_average', 'vote_count', 'popularity', 'keywords', 'overview']]
```

**텍스트 문자 1차 가공. 파이썬 딕셔너리 변환 후 리스트 형태로 변환**

```python
from ast import literal_eval

movie_df['genres'] = movie_df['genres'].apply(literal_eval)
movie_df['keywords'] = movie_df['keywords'].apply(literal_eval)
```

```python
movie_df['genres'].head(1)
```

```python
movie_df['genres'] = movie_df['genres'].apply(lambda x: [i['name'] for i in x])
movie_df['keywords'] = movie_df['keywords'].apply(lambda x: [i['name'] for i in x])
movie_df[['genres', 'keywords']].head(1)
```

**장르 문자열의 Count 기반 피처 벡터화**

```python
(' ').join(['test1', 'test2'])
```

```python
from sklearn.feature_extraction.text import CountVectorizer

movie_df['genres_literal'] = movie_df['genres'].apply(lambda x : (' ').join(x))
count_vect = CountVectorizer(min_df=1, ngram_range=(1, 2))
genre_mat = count_vect.fit_transform(movie_df['genres_literal'])
print(genre_mat.shape)
```

**장르에 따른 영화별 코사인 유사도 추출**

```python
from sklearn.metrics.pairwise import cosine_similarity

genre_sim = cosine_similarity(genre_mat, genre_mat)
print(genre_sim.shape)
print(genre_sim[:2])
```

```python
genre_sim_sorted_ind = genre_sim.argsort()[:, ::-1] # 인덱스 값이 들어가는 2차원 배열
print(genre_sim_sorted_ind[:1])
```

**특정 영화와 장르별 유사도가 높은 영화를 반환하는 함수 생성**

```python
def find_sim_movie(df, sorted_ind, title_name, top_n=10):

    title_movie = df[df['title'] == title_name]

    title_index = title_movie.index.values
    similar_indexes = sorted_ind[title_index, :(top_n)]

    print(similar_indexes)
    similar_indexes = similar_indexes.reshape(-1)

    return df.iloc[similar_indexes]
```

```python
similar_movies = find_sim_movie(movie_df, genre_sim_sorted_ind, 'The Godfather', 10)
similar_movies[['title', 'vote_average']]
```

**평점이 높은 영화 정보 확인**

```python
# vote_count 가 낮은 데이터는 신뢰하기 어렵다 -> 평점만으로는 반영 x
movie_df[['title', 'vote_average', 'vote_count']].sort_values(['vote_average'], ascending=False)[:10]
```

**Weighted Rating (WR)**

WR = $(v / (v + m)) * R + (m / (v + m)) * C$

Where:
- R = 영화의 평균 평점 (average rating)
- v = 해당 영화의 투표 수 (vote count)
- m = 평점을 신뢰할 수 있는 최소 투표 수 (minimum votes required)
- C = 전체 영화의 평균 평점 (mean vote across all movies)

```python
C = movie_df['vote_average'].mean()
m = movie_df['vote_count'].quantile(0.6)

print('C :', round(C, 3), 'm: :',round(m, 3))
```

```python
percentile = 0.6
C = movie_df['vote_average'].mean()
m = movie_df['vote_count'].quantile(percentile)

def weighted_vote_average(record):
    v = record['vote_count']
    R = record['vote_average']
    return ( (v/(v+m)) * R) + ( (m / (v+m)) * C)

movie_df['weighted_vote'] = movie_df.apply(weighted_vote_average, axis=1)
```

```python
movie_df[['title', 'vote_average', 'weighted_vote', 'vote_count']].sort_values('weighted_vote', ascending=False)
```

```python
def find_sim_movie(df, sorted_ind, title_name, top_n=10):
    title_movie = df[df['title'] == title_name]
    title_index = title_movie.index.values

    similar_indexes = sorted_ind[title_index, :(top_n*2)]
    print(similar_indexes)
    similar_indexes = similar_indexes.reshape(-1)
    # 기준 영화 index는 제외
    similar_indexes = similar_indexes[similar_indexes != title_index]

    return df.iloc[similar_indexes].sort_values('weighted_vote', ascending=False)[:top_n]

similar_movies = find_sim_movie(movie_df, genre_sim_sorted_ind, 'The Godfather', 10)
similar_movies[['title', 'vote_average', 'weighted_vote']]
```
