```python
import kagglehub
from sklearn.feature_extraction.text import CountVectorizer

# Download latest version
path = kagglehub.dataset_download("tmdb/tmdb-movie-metadata")

print("Path to dataset files:", path)
```

```python
import pandas as pd
import numpy as np
import warnings; warnings.filterwarnings("ignore")

movies = pd.read_csv(f"{path}/tmdb_5000_movies.csv")
print(movies.shape)
movies.head()
```

```python
movie_df = movies[['id', 'title', 'genres', 'vote_average', 'vote_count', 'popularity', 'keywords', 'overview']]
```

```python
pd.set_option('max_colwidth', 100)
movie_df[['genres', 'keywords']]
```

```python
movie_df.info()
```

**텍스트 문자 1차 가공. 파이썬 딕셔너리 변환 후 리스트 형태로 변환**

```python
from ast import literal_eval

movie_df['genres'] = movie_df['genres'].apply(literal_eval)
movie_df['keywords'] = movie_df['keywords'].apply(literal_eval)
```

```python
movie_df['genres'].head(1)
```

```python
movie_df['genres'] = movie_df['genres'].apply(lambda x: [i['name'] for i in x])
movie_df['keywords'] = movie_df['keywords'].apply(lambda x: [i['name'] for i in x])
movie_df[['genres', 'keywords']].head(1)
```

**장르 문자열의 Count 기반 피처 벡터화**

```python
(' ').join(['test1', 'test2'])
```

```python
from sklearn.feature_extraction.text import CountVectorizer

movie_df['genres_literal'] = movie_df['genres'].apply(lambda x : (' ').join(x))
count_vect = CountVectorizer(min_df=1, ngram_range=(1, 2))
genre_mat = count_vect.fit_transform(movie_df['genres_literal'])
print(genre_mat.shape)
```

**장르에 따른 영화별 코사인 유사도 추출**

```python
from sklearn.metrics.pairwise import cosine_similarity

genre_sim = cosine_similarity(genre_mat, genre_mat)
print(genre_sim.shape)
print(genre_sim[:2])
```

```python
genre_sim_sorted_ind = genre_sim.argsort()[:, ::-1] # 인덱스 값이 들어가는 2차원 배열
print(genre_sim_sorted_ind[:1])
```
