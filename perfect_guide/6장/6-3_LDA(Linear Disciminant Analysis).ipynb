### 붓꽃 데이터 셋에 LDA 적용하기

```python
from sklearn.discriminant_analysis import LinearDiscriminantAnalysis
from sklearn.preprocessing import StandardScaler
from sklearn.datasets import load_iris

iris = load_iris()
iris_scaled = StandardScaler().fit_transform(iris.data)
```

```python
ida = LinearDiscriminantAnalysis(n_components=2)
# fit()호출 시 target값 입력
ida.fit(iris_scaled, iris.target)
iris_ida = ida.transform(iris_scaled)
print(iris_ida.shape)
```

```python
import pandas as pd
import matplotlib.pyplot as plt
%matplotlib inline

lda_colunms = ['lda_component_1', 'lda_component_2']
irisDF_lda = pd.DataFrame(iris_ida, columns=lda_colunms)
irisDF_lda['target'] = iris.target

# setosa는 세모, versicolor는 네모, virginica는 동그라미로 표현
markers = ['^', 's', 'o']

for i, markers in enumerate(markers):
    x_axis_data = irisDF_lda[irisDF_lda['target'] == i]['lda_component_1']
    y_axis_data = irisDF_lda[irisDF_lda['target'] == i]['lda_component_2']
    
    plt.scatter(x_axis_data, y_axis_data, marker=markers, label=iris.target_names[i])

plt.legend()
plt.xlabel('lda_compontent_1')
plt.ylabel('lda_compontent_2')
plt.show()
```
